# Analyse - Modèle Linéaire de Classification

## Objectif
Tester un modèle linéaire sur 3 cas:
1. **Dataset linéaire** - modèle doit réussir
2. **Dataset non-linéaire (XOR)** - modèle linéaire échoue
3. **Transformation non-linéaire** - ajouter features pour résoudre le problème

## Historique des Tests

### Test 1: Dataset Linéaire
- **Description**: Points facilement séparables linéairement
- **Nombre d'échantillons**: 6
- **Résultat**: **Accuracy = 100%** ✓
- **Conclusion**: Le modèle linéaire fonctionne parfaitement sur les données linéaires

### Test 2: Dataset Non-Linéaire (Cas KO - XOR)
- **Description**: Problème classique XOR non séparable linéairement
- **Nombre d'échantillons**: 8
- **Résultat**: **Accuracy = 80%** ✗
- **Conclusion**: Le modèle linéaire simple échoue. C'est le cas KO prévu - linéaire ne suffit pas

### Test 3: Transformation Non-Linéaire
- **Description**: Ajout de features non-linéaires (interaction x1*x2 + r²)
- **Résultat**: **Accuracy = 100%** ✓
- **Conclusion**: En augmentant les dimensions avec des features non-linéaires, le problème devient séparable!

## Architecture du Projet

### Pipeline Complet
```
1. Python (main.py)
   - Redimensionne images → 224×224
   - Compresse intelligemment (JPEG)
   - Extrait features RGB (r_mean, g_mean, b_mean)
   - Génère dataset.csv

2. C (model_c/main.c)
   - Charge dataset.csv
   - Entraîne modèle linéaire
   - Teste sur 3 cas (linéaire, KO, transformation)
   - Affiche accuracy
```

### Modèle Linéaire
```
Equation: y = w1*x1 + w2*x2 + w3*x3 + bias
Classement: pred_class = (y >= 0.5) ? 1 : 0
Entraînement: Descente de gradient simple
Learning rate: 0.01
```

### Transformation Non-Linéaire
Ajout de features:
- **x4** = x1 * x2 (interaction)
- **x5** = x1² + x2² (distance radiale)

## Résumé des Résultats

| Test | Cas | Accuracy | Statut |
|------|-----|----------|--------|
| 1 | Dataset linéaire | 100% | ✓ Pass |
