## 1.1 - kmeam | 24*24 | Gray Scale | 1k n/classe

```python

# CONFIG
IMG_SIZE   = 24
COULEUR    = "gray"
ALGO       = "random"
N_PAR_CLS  = 1000

df = data_set_separation(seed=12, nb_img_per_classes=N_PAR_CLS, ratios=(0.70, 0.15, 0.15))
X_train, X_val, X_test, y_train, y_val, y_test = data_set_transform(df, img_size=IMG_SIZE, color=COULEUR)

dim = X_train.shape[1]
```

```python


# HP
nb_epoch      = 100
learning_rate = .1
seeds         = [3, 8, 9]

gammas        = [.001, .01, .1, .5, .8, 2]

list_nb_centre = np.arange(50, 350, 50)

set_algo(ALGO)

results = {}
for c in list_nb_centre:
    for g in gammas:
        print(f"nb_centre={c} | gamma={g}")
        results[(c, g)] = rbf_classification_training(
                                nb_epoch=nb_epoch,
                                learning_rate=learning_rate,
                                nb_centre=int(c),
                                gamma=g,
                                seeds=seeds,
                                X_train=X_train, X_val=X_val,
                                y_train=y_train, y_val=y_val,
                            )
```

```python
CLASSES = ["compost", "dechets_chimiques", "recyclable"]

ks     = sorted({k for k, g in results})
gs     = sorted({g for k, g in results})
epochs = np.arange(nb_epoch)

# grilles de resultats (lignes = k, colonnes = gamma)
A_final = np.array([[results[(k, g)]["accs_mean"][-1]  for g in gs] for k in ks])
A_best  = np.array([[results[(k, g)]["accs_mean"].max() for g in gs] for k in ks])
A_std   = np.array([[results[(k, g)]["accs_std"][-1]   for g in gs] for k in ks])

# couple retenu (sur accuracy finale)
i_star, j_star = np.unravel_index(np.argmax(A_final), A_final.shape)
k_star, g_star = ks[i_star], gs[j_star]
r_star = results[(k_star, g_star)]

fig, ((axA, axB), (axC, axD), (axE, axF)) = plt.subplots(3, 2, figsize=(18, 17))
fig.suptitle(f"RBF {ALGO} | {IMG_SIZE}x{IMG_SIZE} {COULEUR} | {N_PAR_CLS} img/classe | "
             f"dim={dim} | {len(seeds)} seeds x {nb_epoch} epochs", fontsize=14)

# --- A : heatmap k x gamma (LE graphe de la grille) ---
im = axA.imshow(A_final, cmap="viridis", aspect="auto", origin="lower")
axA.set_xticks(range(len(gs))); axA.set_xticklabels(gs)
axA.set_yticks(range(len(ks))); axA.set_yticklabels(ks)
axA.set_xlabel("gamma (relatif)"); axA.set_ylabel("nb_centre (k)")
axA.set_title("Accuracy validation (derniere epoch)")
thr = (A_final.max() + A_final.min()) / 2
for a in range(len(ks)):
    for b in range(len(gs)):
        axA.text(b, a, f"{A_final[a,b]:.3f}", ha="center", va="center", fontsize=8,
                 color="white" if A_final[a,b] < thr else "black")
axA.plot(j_star, i_star, "r*", markersize=22, markeredgecolor="white")
fig.colorbar(im, ax=axA)

# --- B : choix de k (une courbe par gamma) ---
for b, g in enumerate(gs):
    axB.plot(ks, A_final[:, b], "o-", label=f"gamma={g}")
axB.axvline(k_star, color="red", linestyle=":", label=f"k* = {k_star}")
axB.set_xlabel("nb_centre (k)"); axB.set_ylabel("accuracy validation")
axB.set_title("Effet de k"); axB.legend(fontsize=8); axB.grid(True)

# --- C : choix de gamma (une courbe par k) ---
for a, k in enumerate(ks):
    axC.plot(gs, A_final[a, :], "o-", label=f"k={k}")
axC.axvline(g_star, color="red", linestyle=":", label=f"gamma* = {g_star}")
axC.set_xscale("log")
axC.set_xlabel("gamma (relatif, echelle log)"); axC.set_ylabel("accuracy validation")
axC.set_title("Effet de gamma"); axC.legend(fontsize=8); axC.grid(True, which="both")

# --- D : accuracy par epoch a gamma* (regle nb_epoch) ---
for k in ks:
    r = results[(k, g_star)]
    axD.plot(epochs, r["accs_mean"], label=f"k={k}")
    axD.fill_between(epochs, r["accs_mean"] - r["accs_std"],
                     r["accs_mean"] + r["accs_std"], alpha=0.12)
axD.set_xlabel("epoch"); axD.set_ylabel("accuracy")
axD.set_title(f"Accuracy validation par epoch (gamma*={g_star})")
axD.legend(fontsize=8); axD.grid(True)

# --- E : loss train/val au couple retenu (detecte l'overfit) ---
axE.plot(epochs, r_star["train_loss_mean"], label="train")
axE.fill_between(epochs, r_star["train_loss_mean"] - r_star["train_loss_std"],
                 r_star["train_loss_mean"] + r_star["train_loss_std"], alpha=0.15)
axE.plot(epochs, r_star["val_loss_mean"], "--", label="val")
axE.fill_between(epochs, r_star["val_loss_mean"] - r_star["val_loss_std"],
                 r_star["val_loss_mean"] + r_star["val_loss_std"], alpha=0.15)
axE.axvline(int(np.argmin(r_star["val_loss_mean"])), color="grey", linestyle=":",
            label=f"min val loss (ep {int(np.argmin(r_star['val_loss_mean']))})")
axE.set_xlabel("epoch"); axE.set_ylabel("loss (MSE)")
axE.set_title(f"Loss au couple retenu (k*={k_star}, gamma*={g_star})")
axE.legend(); axE.grid(True)

# --- F : matrice de confusion au couple retenu ---
model_star = r_star["model"]        # /!\ modele du DERNIER seed uniquement
cm = confusion_matrix(y_val, model_star.predict(X_val), 3)
cm_pct = cm / cm.sum(axis=1, keepdims=True)

im = axF.imshow(cm_pct, cmap="Blues", vmin=0, vmax=1)
axF.set_xticks(range(3)); axF.set_xticklabels(CLASSES, rotation=45, ha="right")
axF.set_yticks(range(3)); axF.set_yticklabels(CLASSES)
axF.set_xlabel("Predit"); axF.set_ylabel("Vrai")
axF.set_title(f"Confusion (Val) - k*={k_star}, gamma*={g_star}\n"
              f"Accuracy = {A_final[i_star, j_star]*100:.1f}%  (seed {seeds[-1]})")
for a in range(3):
    for b in range(3):
        axF.text(b, a, f"{int(cm[a,b])}\n{cm_pct[a,b]*100:.0f}%", ha="center", va="center",
                 color="white" if cm_pct[a,b] > 0.5 else "black")
fig.colorbar(im, ax=axF)

plt.tight_layout()
plt.show()

# ==== LIGNE DU TABLEAU ====
print(f"k opti        = {k_star}")
print(f"gamma opti    = {g_star}   (relatif ; absolu = {g_star/dim:.2e})")
print(f"acc finale    = {A_final[i_star, j_star]:.4f} +/- {A_std[i_star, j_star]:.4f}")
print(f"acc max       = {A_best[i_star, j_star]:.4f}")
print(f"epoch optimal = {int(np.argmax(r_star['accs_mean']))} / {nb_epoch}")
```

## Pourquoi Kmean++ perform moins bien a centre egal et a besoin de plus de centre ?

```python
rng = np.random.default_rng(1)
K = 12

# --- Données : 4 amas + quelques outliers ---
amas    = np.array([[-4, -4, 0], [4, -4, 0], [0, 4, -3], [0, 0, 4]], float)
X_dense = np.vstack([c + rng.normal(0, .8, (150, 3)) for c in amas])
X_out   = rng.uniform(-11, 11, (30, 3))
X       = np.vstack([X_dense, X_out])

def d2(A, B):
    return np.maximum((A**2).sum(1)[:, None] + (B**2).sum(1)[None, :] - 2 * A @ B.T, 0)

def init_naif(X, K, rng):        # tirage uniforme
    return X[rng.choice(len(X), K, replace=False)]

def init_kpp(X, K, rng):         # tirage en D²
    idx = [rng.integers(len(X))]
    d = d2(X, X[idx]).ravel()
    for _ in range(K - 1):
        i = rng.choice(len(X), p=d / d.sum())
        idx.append(i)
        d = np.minimum(d, d2(X, X[[i]]).ravel())
    return X[idx]

fig, axes = plt.subplots(1, 2, figsize=(11, 5.5))

for ax, init, nom in [(axes[0], init_naif, "k-means naïf"),
                      (axes[1], init_kpp,  "k-means++")]:
    C = init(X, K, np.random.default_rng(1))
    outlier = d2(C, amas).min(1) > 25       # loin de tout amas -> centre sur un outlier

    ax.scatter(X[:, 0], X[:, 1], c="lightgrey", s=8)
    ax.scatter(C[~outlier, 0], C[~outlier, 1], s=90, edgecolors="k")               # amas
    ax.scatter(C[outlier, 0],  C[outlier, 1], marker="X", s=150, c="red", edgecolors="k")   # outliers
    ax.set_title(nom)
    ax.set_xticks([]); ax.set_yticks([])

plt.tight_layout()
plt.show()
```
