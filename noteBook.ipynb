```python
import numpy as np
import matplotlib.pyplot as plt
from rbf_main import RBF, one_hot, set_seed
```

```python
X = np.array([[1,1],[2,3],[3,3]], float)
y = np.array([0, 1, 1])

set_seed(1)
model = RBF(nb_centres=3, gamma=0.5)
mse = model.train(X, one_hot(y, 2))      # UN seul appel
print(f"MSE finale : {mse:.6f}")
```

```python
gammas = np.linspace(0.05, 5, 30)
mses = []
for g in gammas:
    m = RBF(nb_centres=3, gamma=g)
    val = m.train(X, one_hot(y, 2))
    mses.append(val if val >= 0 else np.nan)   # -1 = matrice singuliere

plt.plot(gammas, mses, "-o", ms=3)
plt.xlabel("gamma"); plt.ylabel("MSE")
plt.title("RBF : MSE en fonction de gamma")
plt.grid(True); plt.show()
```

```python
xx, yy = np.meshgrid(np.linspace(0, 4, 300), np.linspace(0, 4, 300))
grid = np.c_[xx.ravel(), yy.ravel()]

scores = np.array([model.forward(pt) for pt in grid])   # comme ton PMC
Z = scores.argmax(axis=1).reshape(xx.shape)

plt.contourf(xx, yy, Z, alpha=0.4, cmap="coolwarm")
plt.scatter(X[y==0,0], X[y==0,1], color="blue", s=100, zorder=5)
plt.scatter(X[y==1,0], X[y==1,1], color="red",  s=100, zorder=5)
plt.title("Frontière de décision - RBF")
plt.show()
```
